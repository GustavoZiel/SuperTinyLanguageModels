```python
import string


def extract_keys(fmt: str):
    formatter = string.Formatter()
    return [field_name for _, field_name, _, _ in formatter.parse(fmt) if field_name]


# Example
s = "{NAME} was born on {BIRTH_DATE}"
print(extract_keys(s))
# → ['NAME', 'BIRTH_DATE']
```
